### Q2.

Dado um número inteiro positivo informado pelo usuário, crie um programa em Python que utilize o comando for para calcular e exibir a soma dos números de 1 até o número informado. Ou seja, $1 + 2 + 3 + ... + n$

Exemplo de interação:
```
Digite um número: 5
A soma dos números de 1 a 5 é 15
```


````python
N = int(input("Digite um número: "))
soma = 0
for i in range(1, N + 1):
    soma += i
print(f"A soma dos números de 1 a {N} é: {soma}")
````

### Q3.

Escreva um programa que lê 10 valores inteiros positivos do usuário e ao final imprime a média dos valores digitados com duas casas decimais.

```
Digite 10 números positivos:
3
2
12
14
20
42
19
0
8
1
A média dos valores digitados é 12.10
```

````python
# Entrada de dados do usuário. Digitar 10 números inteiros e calcular a média aritmética deles.
soma = 0
for i in range(10):
    num = int(input("Digite um número natural: "))
    print(num)
    soma += num
print(f"A soma dos números digitados é: {soma}")
print(f"A média dos números digitados é:\033[1;32m{soma / 10: .2f}\033[m")
````

### Q4.

Você vai criar um sistema que registra os resultados dos jogos do Atlético MG ao longo de um campeonato. Seu sistema vai receber os resultados de todos os jogos do Galo, e deve calcular a pontuação do time sabendo que vitórias valem 3 pontos, empates 1 ponto e derrotas 0 pontos.

Entrada: <br>
A primeira linha de entrada é um inteiro N com a quantidade jogos do galo. Para cada jogo você deve ler 2 inteiros, o primeiro com a quantidade de gols do galo e o segundo com a quantidade de gols do time oponente. 

Saída: <br>
Apresente a soma de vitórias, empates e derrotas do galo, junto com o cálculo da pontuação total.

Ex:
```
Entrada:
4
2
0
1
1
0
3
1
0

Saída:
Vitórias: 2
Empates: 1
Derrotas: 1
Pontuação: 7

```




````python
# Entrada de dados. Jogos do Galo, gols feitos e tomados e pontuação total no campeonato.

N = int(input("Digite a quantidade de jogos do Galo, no campeonato: "))
print(f"Quantidade de jogos do Galo no campeonato: \033[1;35m{N}\033[0m")

vitorias, empates, derrotas = 0, 0, 0

for jogo in range(1, N + 1):
    gols_Galo = int(input(f"Digite o número de gols do Galo no jogo {jogo} "))
    print(f"Total de gols do \033[1;31mGalo\033[0m no \033[1;35mjogo {jogo}\033[0m: \033[1;33m{gols_Galo}\033[0m")
    gols_adversario = int(input(f"Digite o número de gols do adversário no jogo {jogo} "))
    print(f"Total de gols do \033[1;31mAdversário\033[0m no \033[1;35mjogo {jogo}\033[0m: \033[1;33m{gols_adversario}\033[0m")
    if gols_Galo > gols_adversario:
        vitorias += 1
    elif gols_Galo < gols_adversario:
        derrotas += 1
    else:
        empates += 1

print(f"Total de \033[1;32mvitórias\033[0m do Galo: \033[1;32m{vitorias}\033[0m")
print(f"Total de \033[1;36mempates\033[0m do Galo: \033[1;36m{empates}\033[0m")
print(f"Total de \033[1;31mderrotas\033[0m do Galo: \033[1;31m{derrotas}\033[0m")
print(f"Pontuação total do Galo no campeonato: \033[1;33m{vitorias * 3 + empates}\033[0m pontos")
````

### Q5.

Faça um programa que lê dois inteiros N e M, e imprime na tela um campo de batalha naval. O tabuleiro deve possuir N linhas e M colunas. A primeira linha é composta por um espaço em branco e o cabeçalho das colunas, ou seja, valores de 1 a M. As N linhas seguintes iniciam com o cabeçalho da linha, ou seja, seu número, seguido de M caracteres "/" (barra) indicando uma possível posição jogável. 

```
Entrada:
5
4

Saída:
  1 2 3 4 
1 / / / / 
2 / / / / 
3 / / / / 
4 / / / / 
5 / / / / 
```

Para esse exercício, precisamos lembrar que o comando print implicitamente adiciona uma quebra de linha ao final da impressão. Podemos interferir no final da impressão adicionando mais uma entrada ao print. No exemplo, finalizamos cada linha com um espaço em branco:
```python
print("Texto qualquer", end = " ")
```

````python
# Criar uma cartela do jogo Batalha Naval. Entrada de dados, número de linhas e colunas, e imprimir a cartela.
N = int(input("Digite o número de linhas da cartela: "))
M = int(input("Digite o número de colunas da cartela: "))

for i in range(1, M+1):
    print("\033[1;42m ", i, end=" \033[0m")
for i in range(1, N + 1):
    print(f"\033[1;42m\n{i}", end=" \033[0m")
    for j in range(1, M + 1):
        print("\033[1;44m / ", end=" \033[0m")
````

### Q6.

Escreva um programa que lê uma quantidade indefinida de valores e informa o maior e o menor valor digitados. A leitura é encerrada quando o usuário digitar o valor 0 (zero).

```
Entrada:
12
-1
29
11
-5
0

Saída:
Maior: 29
Menor: -5
```

````python
maior = None
menor = None

while True:
    n = int(input("Digite um número Inteiro (positivo, negativo ou zero): "))
    print(f"Você digitou o número: {n}")

    if n == 0:
        break
              
    if maior is None or n > maior:
            maior = n
    if menor is None or n < menor:
            menor = n
        
print(f"O maior número digitado foi: {maior}")
print(f"O menor número digitado foi: {menor}")
````

### Q7.

Escreva um programa que lê uma quantidade indefinida de valores e informa o **produto** dos valores positivos digitados. A leitura é encerrada quando o usuário digitar o valor 0 (zero).

```
Entrada:
2
-1
1
3
-5
0

Saída:
Produto: 6
```

````python
produto = 1
while True:
    n = int(input("Digite um número inteiro positivo (ou zero para sair): "))
    print(f"Você digitou o número: {n}")
    if n == 0:
        break
    elif n < 0:
        print("Número inválido. Digite um número positivo.")
        continue
    produto *= n
print(f"O produto dos números digitados é: {produto}")

````

### Q8.

Vamos fazer uma calculadora que aceita expressões aritméticas de qualquer tamanho até que o usuário digite a palavra "Fim". Sua calculadora executa apenas as operações "+", "-".

```
Entrada:
5
+
8
-
3
+
2
Fim

Saída:
12
```

````python
resultado = int(input("Digite um número inteiro positivo: "))
print(f"Você inseriu o número: {resultado}")

while True:
    operacao = input("Digite a operação desejada (+, -) ou 'Fim' para encerrar: ")
    print(f"Você escolheu a operação: {operacao}")
    if operacao == "Fim":
        break
    n = int(input("Digite o próximo número inteiro positivo: "))
    print(f"Você inseriu o número: {n}")
    if operacao == "+":
        resultado += n
    elif operacao == "-":
        resultado -= n
    else:
        print("Operação inválida. Digite apenas '+', '-' ou 'Fim'.")
                  

print(f"O resultado da operação é: {resultado}")
````
